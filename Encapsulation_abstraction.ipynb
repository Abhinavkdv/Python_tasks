```python
# 1
class BankAccount:

    def __init__(self,bal):
        self.__balance=bal

    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount
        else:
            print("enter a positive integer value !!!")

    def withdraw(self,amount):
        if amount>0 and amount<=self.__balance:
            self.__balance-=amount
            print("amount is withdrawed")
        elif amount>self.__balance:
            print("you balance is low")
        else:
            print("enter a positive integer value !!!")

ba = BankAccount(20_000)
ba.deposit(10_000)
ba.withdraw(15_000)
```

```python
# 16

class Shape(ABC):

    @abstractmethod
    def calculate_area(self):
        pass

    @abstractmethod
    def calculate_perimeter(self):
        pass

class Circle(Shape):
    
    def __init__(self,rad):
        self.r=rad
    
    def calculate_area(self):
        print(f"Area of the circle is {(3.14)*self.r*self.r}")

    def calculate_perimeter(self):
        print(f"perimeter of the circle is {(3.14)*2*self.r}")

class Rectangle(Shape):
    
    def __init__(self,L,B):
        self.l=L
        self.b=B
    
    def calculate_area(self):
        print(f"Area of the Rectangle is {self.l*self.b}")

    def calculate_perimeter(self):
        print(f"perimeter of the Rectangle is {2*(self.l+self.b)}")

cir = Circle(5)
rec = Rectangle(5,4)

cir.calculate_area()
cir.calculate_perimeter()

rec.calculate_area()
rec.calculate_perimeter()
```

```python
# 25
class Shape3D(ABC):

    @abstractmethod
    def volume(self):
        pass

    @abstractmethod
    def surface_area(self):
        pass

class Cube(Shape3D):

    def __init__(self,s):
        self.side=s

    def volume(self):
        print(f"volume of the cube is {self.side**3}")

    def surface_area(self):
        print(f"surface area of the cube is {6*(self.side**2)}")

class Sphere(Shape3D):

    def __init__(self,s):
        self.r=s

    def volume(self):
        print(f"volume of the Sphere is {(3.14)*(4/3)*(self.r**3)}")

    def surface_area(self):
        print(f"surface area of the Sphere is {4*(3.14)*(self.r**2)}")

class Cylinder(Shape3D):

    def __init__(self,R,H):
        self.r=R
        self.h=H
        

    def volume(self):
        print(f"volume of the Cylinder is {(3.14)*(self.r**2)*(self.h)}")

    def surface_area(self):
        print(f"surface area of the Cylinder is {2*(3.14)*(self.r)*(self.r+self.h)}")

cube = Cube(4)
sphere = Sphere(4)
cylinder = Cylinder(2,3)

cube.volume()
cube.surface_area()

sphere.volume()
sphere.surface_area()

cylinder.volume()
cylinder.surface_area()
```

```python
# 27
class Database(ABC):
    
    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def query(self,sql):
        pass

class MySQLDatabase(Database):

    def connect(self):
        print("Your connected to MySQLDatabase...")

    def query(self,sql):
        print(f"Query: {sql}")

class MongoDatabase(Database):

    def connect(self):
        print("Your connected to MongoDatabase...")

    def query(self,sql):
        print(f"Query: {sql}")

sql = MySQLDatabase()
md = MongoDatabase()

sql.connect()
sql.query("Select * from Table")
print("\n")
md.connect()
md.query(" db.collection.find()")
```

```python
# 29
class Transport(ABC):

    @abstractmethod
    def move(self):
        pass

    @abstractmethod
    def fuel_type(self):
        pass

class Car(Transport):

    def move(self):
        print("Car moves")

    def fuel_type(self):
        print("Petrol")

class Train(Transport):

    def move(self):
        print("Train moves")

    def fuel_type(self):
        print("Electricity")

class Airplane(Transport):

    def move(self):
        print("Airplane moves")

    def fuel_type(self):
        print("Aviation Fuel")

car = Car()
train = Train()
airplane = Airplane()

car.move()
car.fuel_type()
print("\n")
train.move()
train.fuel_type()
print("\n")
airplane.move()
airplane.fuel_type()
```

```python
# 30
class FileHandler(ABC):

    @abstractmethod
    def read(self):
        pass

    @abstractmethod
    def write(self,data):
        pass

class TextFileHandler(FileHandler):

    def read(self):
        print("Reading textfile")

    def write(self,data):
        print("writing....")
        print(data)

class CSVFileHandler(FileHandler):

    def read(self):
        print("Reading CSV file")

    def write(self,data):
        print("writing....")
        print(data)

tx = TextFileHandler()
csv = CSVFileHandler()

tx.read()
tx.write("text")
print("\n")
csv.read()
csv.write("text")
```
