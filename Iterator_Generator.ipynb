```python
# 27
class prime:

    def __init__(self,num):
        self.i =2
        self.num = num

    
    def __iter__(self):
        return self

    def __next__(self):
        j=2
        count =0
        while j<self.i:
            if self.i%j==0:
                count+=1
            j+=1
        if count==0:
            x = self.i
            return x
        self.i+=1
        raise StopIteration

itr = prime(100)
print(next(itr))
print(next(itr))
print(next(itr))
print(next(itr))
```

```python
# 28
class factor:
    def __init__(self,num):
        self.i =1
        self.num = num

    
    def __iter__(self):
        return self

    def __next__(self):
        while self.i<=self.num:
            if self.num%self.i==0:
                x = self.i
                self.i+=1
                return x
            self.i+=1
        raise StopIteration

itr = factor(24)
print(next(itr))
print(next(itr))
print(next(itr))
print(next(itr))
print(next(itr))
print(next(itr))
print(next(itr))
print(next(itr))
```

```python
# 29
def countdown(num):
    i=num
    while i>0:
        yield i
        i=i-1

gen = countdown(20)
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```
