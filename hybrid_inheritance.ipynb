```python
# 1
class A:
    def methodA(self):
        print("This method is from class A")

class B(A):
    def methodB(self):
        print("This method is from class B")

class C:
    def methodC(self):
        print("This method is from class C")

class D(B,C):
    def methodD(self):
        print("This method is from class D")

obj = D()
obj.methodA()
obj.methodB()
obj.methodC()
obj.methodD()
```

```python
# 2
class Person:
    def __init__(self,name,**kwargs):
        self.__name=name

    def get_name(self):
        return self.__name

class Employee(Person):
    def __init__(self,name,salary,**kwargs):
        super().__init__(name,**kwargs)
        self.__salary=salary

    def get_salary(self):
        return self.__salary

class Student(Person):
    def __init__(self,name,marks,**kwargs):
        super().__init__(name)
        self.__marks=marks

    def get_marks(self):
        return self.__marks

class TeachingAssistant(Employee,Student):
    def __init__(self,name,salary,marks,department):
        super().__init__(
            name=name,
            salary=salary,
            marks=marks
        )
        self.department=department

    def get_department(self):
        return self.department

    def display(self):
        print("Name:",self.get_name())
        print("Salary:",self.get_salary())
        print("Marks:",self.get_marks())
        print("Department:",self.get_department())

ta = TeachingAssistant("Abhinav",20_000,89,"CSE")
ta.display()
```

```python
# 3
class Vehicle:
    def start(self):
        print("This is start method")

class Car(Vehicle):
    def drive(self):
        print("This is drive method")

class Electric(Vehicle):
    def battery(self):
        print("This is battery method")

class ElectricCar(Car,Electric):
    pass

el = ElectricCar()
el.start()
el.drive()
el.battery()
```

```python
# 5
class A:
    def __init__(self,name):
        print("Constructor called!!!")
        self.name=name
        
    def get_name(self):
        return self.name

class B(A):
    def __init__(self,name):
        super().__init__(name)

class C(A):
    def __init__(self,name):
        super().__init__(name)

class D(B,C):
    def __init__(self,name):
        super().__init__(name)

d = D("Abhinav")
print(d.get_name())
```

```python
# 6
class Account:
    def __init__(self,account_no,**kwargs):
        self.__account_no=account_no
        super().__init__(**kwargs) # to pass the remaining arguments to tje mro chain

    def get_account_no(self):
        return self.__account_no

class Savings:
    def __init__(self,balance,**kwargs):
        self.balance=balance
        super().__init__(**kwargs) # to pass the remaining arguments to tje mro chain

class Loan:
    def __init__(self,loan_amt,**kwargs):
        self.loan_amt=loan_amt
        super().__init__(**kwargs) # to pass the remaining arguments to tje mro chain


class SavingsLoanAccount(Account,Savings,Loan):
    def __init__(self,account_no,balance,loan_amt):
        super().__init__(
            account_no=account_no,
            balance=balance,
            loan_amt=loan_amt
        )

    def display(self):
        print("Account Number:",self.get_account_no())
        print("Balance:",self.balance)
        print("Loan Amount:",self.loan_amt)

obj = SavingsLoanAccount(123456789,50_000,1_00_000)
obj.display()
```

```python
# 8
class Electronics:
    def __init__(self,voltage,**kwargs):
        print("Electronics Constructor")
        self.voltage=voltage
        super().__init__(**kwargs)

class Audio:
    def __init__(self,volume,**kwargs):
        print("Audio Constructor")
        self.volume=volume
        super().__init__(**kwargs)

class WiFi:
    def __init__(self,ssid,**kwargs):
        print("WiFi Constructor")
        self.ssid=ssid
        super().__init__(**kwargs)

class SmartSpeaker(Electronics,Audio,WiFi):
    def __init__(self,voltage,volume,ssid):
        super().__init__(
            voltage = voltage,
            volume = volume,
            ssid = ssid)

ss = SmartSpeaker("5kW",100,1)
print(SmartSpeaker.mro())
```

```python
# 15
class Circle:
    
    def __init__(self,r):
        self.r=r
        
    def area(self):
        return ((self.r**2)*3.14)

class Rectangle:
    
    def __init__(self,l,b):
        self.l=l
        self.b=b
        
    def area(self):
        return self.l*self.b

def calculate(obj):
    print(obj.area())

calculate(Circle(5))
calculate(Rectangle(5,4))
```

```python
# 19
class GrandParent:
    pass
    
class ParentA(GrandParent):
    pass

class ParentB:
    pass

class HybridChild(ParentA,ParentB):
    def show(self):
        print("This is from Hybrid child")

class UnrelatedClass:
    def show(self):
        print("This is from UnrelatedClass ")

def execute_show(obj):
    obj.show()

execute_show(HybridChild())
execute_show(UnrelatedClass())
print(HybridChild.mro())
print(UnrelatedClass.mro())
```

```python
# 20
class Person:
    pass
    
class Employee(Person):
    pass

class Manager(Employee):
    pass

class HR(Employee):
    pass

class TeamLead(Manager,HR):
    def work(self):
        print("Teamlead working...")

class Freelancer:
    def work(self):
        print("Freelancer is working...")

def do_work(obj):
    obj.work()

do_work(TeamLead())
do_work(Freelancer())
print(TeamLead.mro())
print(Freelancer.mro())
```
