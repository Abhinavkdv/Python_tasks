```python
# 6
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited: {amount}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawn: {amount}")
        else:
            print("Insufficient balance.")

class SavingsAccount(BankAccount):
    def calculate_interest(self, rate):
        interest = self.balance * rate / 100
        print(f"Interest: {interest}")

account = SavingsAccount(10000)
account.deposit(2000)
account.withdraw(1500)
account.calculate_interest(5)
print("Current Balance:", account.balance)
```

```python
# 19
class Shape:
    def show(self):
        print("This is a shape.")

class Polygon(Shape):
    def sides(self):
        print("A polygon has multiple sides.")

class Rectangle(Polygon):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        area = self.length * self.width
        print("Area of Rectangle:", area)

rect = Rectangle(10, 5)
rect.show()
rect.sides()
rect.calculate_area()
```

```python
# 28
class Employee:
    def __init__(self, name):
        self.name = name

class Manager(Employee):
    def __init__(self, name, department):
        super().__init__(name)
        self.department = department

class ProjectManager(Manager):
    def __init__(self, name, department, team_size):
        super().__init__(name, department)
        self.team_size = team_size

    def manage_team(self):
        print(f"{self.name} is managing a team of {self.team_size} members.")

    def display(self):
        print("Name:", self.name)
        print("Department:", self.department)
        print("Team Size:", self.team_size)

pm = ProjectManager("Ravi", "IT", 10)
pm.display()
pm.manage_team()
```

```python
# 29
class Shape:
    pass

class TwoDShape(Shape):
    pass

class Triangle(TwoDShape):
    def __init__(self, side1, side2, side3):
        self.side1 = side1
        self.side2 = side2
        self.side3 = side3

    def calculate_area(self):
        s = (self.side1 + self.side2 + self.side3) / 2
        area = (s * (s - self.side1) * (s - self.side2) * (s - self.side3))**0.5
        print("Area of Triangle:", area)

tri = Triangle(3, 4, 5)
tri.calculate_area()
```
