```python
# 13
try:
    file = open("text.txt","r")
    f = file.readlines()
    rev = open("rev_text.txt","w")
    for i in f:
        words = i.split()
        words.reverse()
        for j in words:
            rev.write(j[::-1]+" ")
        rev.write("\n")

    rev.close()
    file.close()
except Exception as e:
    print(e)
```

```python
# 25
try:
    
    file = open("mpc_students_marks.csv","r")
    file1 = open("mpc_result.txt","w")
    
    for i in file:
        if i != "Name,RollNum,Maths,Physics,Chemistry":
            continue
        name,roll_num,m,p,c = i.strip().split(",")
        avg = (int(m) + int(p) + int(c)) / 3
        file1.write(f"Name: {name} Roll Number: {roll_num} Average: { avg }")
        file1.write("\n")
    file1.close()
    file.close()

except Exception as e:
    print(e)  
```
