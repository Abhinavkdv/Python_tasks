```python
# 3
class Teacher:
    def __init__(self,subject,*args):
        super().__init__(*args)
        self.subject=subject
    
    def teaching(self):
        print(f"Explaining topics on {self.subject}")

class Author:
    def __init__(self,bname,*args):
        super().__init__(*args)
        self.bname=bname
    
    def writing(self):
        print(f"writing a Book named {self.bname}")

class Person(Teacher,Author):
    def __init__(self,bname,subject,name):
        super().__init__(bname,subject)
        self.name=name

    
    def display_methods(self):
        print(f"Name is {self.name}")
        self.teaching()
        self.writing()

p = Person("ONE PIECE","Maths","Abhinav")
p.display_methods()
```

```python
# 10

class Father:
    def __init__(self,f_name,f_age,**kwargs):
        super().__init__(**kwargs)
        self.f_name=f_name
        self.f_age=f_age

class Mother:
    def __init__(self,m_name,m_age,**kwargs):
        super().__init__(**kwargs)
        self.m_name=m_name
        self.m_age=m_age

class Child(Father,Mother):
    
    def __init__(self,f_name,f_age,m_name,m_age):
        super().__init__(
            f_name = f_name,
            f_age=f_age,
            m_name = m_name,
            m_age=m_age
        )

    def display(self):
        print(f"Father name: {self.f_name}")
        print(f"Father age: {self.f_age}")
        print(f"Mother name: {self.m_name}")
        print(f"Mother age: {self.m_age}")

child = Child("Ram",42,"Sita",40)
child.display()
print(Child.__mro__)
```

```python
# 11
class MathOperations:

    def __init__(self,a,b,**kwargs):
        super().__init__(**kwargs)
        self.a=a
        self.b=b

    @staticmethod
    def calculate_sum(self):
        print(self.a+self.b)


class DisplayOperations:

    def __init__(self,s,**kwargs):
        super().__init__(**kwargs)
        self.s=s

    @staticmethod
    def message(self):
        print(f"sent a message: '{self.s}' ")

class Phone(MathOperations,DisplayOperations):

    def __init__(self,a,b,s):
        super().__init__(
            a=a,
            b=b,
            s=s
        )

    def display(self):
        self.calculate_sum(self) # why self is needed as a parameter here --> because it is a static method
        self.message(self)

p = Phone(3,4,"hello")
p.display()
```

```python
# 13
class Validator:
    
    @staticmethod 
    def is_valid(age):
        return age>=18
        
class Form(Validator):
    
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def eligible(self):
        if Validator.is_valid(self.age):
            print(f"{self.name} is eligibel")
        else:
            print(f"{self.name} is eligibel")

f= Form("Abhinav",21)
f.eligible()
```

```python
# 14
class Season:
    
    @staticmethod
    def display(season):
        print("Season:",season)
class Converter:
    
    @staticmethod
    def to_celsius(F):
        return (F-32)*(5/9)

class WeatherApp(Converter,Season):

    def __init__(self,F,season):
        self.F = F
        self.season=season

    def display_status(self):
        self.display(self.season)
        print(self.to_celsius(self.F))

wa = WeatherApp(110,"Summer")
wa.display_status()
```

```python
# 15
import math
class Logger:
    
    @staticmethod 
    def log(x):
        return math.log(x)

class MathAdd:
    
    @staticmethod 
    def add(a,b):
        return(a+b)

class Application(Logger,MathAdd):

    def __init__(self,x,a,b):
        self.x=x
        self.a=a
        self.b=b

    def display(self):
        print(self.log(self.x))
        print(self.add(self.a,self.b))

app = Application(2,3,4)
app.display()
```

```python
# 17
class Calculator:

    @staticmethod
    def add(a,b):
        return a+b

class Display:

    @staticmethod
    def display(s):
        print(s)

class A(Calculator,Display):

    def __init__(self,a,b,s):
        self.a=a
        self.b=b
        self.s=s

    def show(self):
        print(self.add(self.a,self.b))
        self.display(self.s)
a = A(3,2,"hello")
a.show()
```

```python
# 19
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def draw(self):
        pass

class Color(ABC):

    @abstractmethod
    def fill_color(self):
        pass

class Circle(Shape, Color):

    def draw(self):
        print("Drawing a Circle")

    def fill_color(self):
        print("Color: Red")

obj = Circle()
obj.draw()
obj.fill_color()
```

```python
# 20
class Payment(ABC):

    @abstractmethod
    def make_payment(self):
        pass

class Notification(ABC):

    @abstractmethod
    def send_notification(self):
        pass

class OnlinePayment(Payment, Notification):

    def make_payment(self):
        print("Payment completed successfully.")

    def send_notification(self):
        print("Notification sent to customer.")

obj = OnlinePayment()
obj.make_payment()
obj.send_notification()
```

```python
# 21
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

class Electric(ABC):

    @abstractmethod
    def charge(self):
        pass

class ElectricCar(Vehicle, Electric):

    def start(self):
        print("Electric car started.")

    def charge(self):
        print("Battery is charging.")

car = ElectricCar()
car.start()
car.charge()
```

```python
# 22
class Readable(ABC):

    @abstractmethod
    def read(self):
        pass

class Writable(ABC):

    @abstractmethod
    def write(self):
        pass

class File(Readable, Writable):

    def read(self):
        print("Reading from file.")

    def write(self):
        print("Writing to file.")

f = File()
f.read()
f.write()
```

```python
# 23
class Flyable(ABC):

    @abstractmethod
    def fly(self):
        pass

class Swimmable(ABC):

    @abstractmethod
    def swim(self):
        pass

class Duck(Flyable, Swimmable):

    def fly(self):
        print("Duck is flying.")

    def swim(self):
        print("Duck is swimming.")

d = Duck()
d.fly()
d.swim()
```
