```python
import copy
```

```python
# 1
original = [1, 2, [3, 4], 5]
shallow = copy.copy(original)
print("shalow list : ",shallow)
shallow[2][0]=99
print("original after update: ",original)
print("shallow after update: ",shallow)
```

```python
# 2
original = [1, 2, [3, 4], 5]
deep = copy.deepcopy(original)
print("deep list : ",deep)
deep[2][0]=99
print("original after update: ",original)
print("deep after update: ",deep)
```

```python
# 8
data = [10, [20, 30], {"name": "Ram"}, 40]
shallow = copy.copy(data)
deep = copy.deepcopy(data)
print("befor update data:",data)
data[1][0] = 15
data[2]["name"] = "Abhi"
print("After update data:",data)
print("shallow copy: ",shallow)
print("deep copy: ",deep)
```

```python
# 9
import copy
def check(lst,data):
    temp = data[0][0]
    if data[0][0] == 9:
        data[0][0] = 1
    else:
        data[0][0]=9
    if data[0][0] == lst[0][0]:
        print("a shallow copy")
    else:
        print("a Deep copy")
    data[0][0] = temp
data = [[1,2],[3,4]]
lst = copy.copy(data)
check(lst,data)
lst2 = copy.deepcopy(data)
check(lst2,data)
```

```python
# 10

class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks

name = ["A","B","C","D"]
marks = [80,90,75,60]
obj = Student(name,marks)
deep = copy.deepcopy(obj)
shallow = copy.copy(obj)
print("Before update original: ",obj.name)
print("marks: ",obj.marks)
shallow.name[0]='E'
print("\nAfter Update a name")
print("Shallow: ",shallow.name)
print("Original: ",obj.name)
print("Deep: ",deep.name)

print("\nAfter updating marks:")
obj.marks[3]=100
print("Shallow: ",shallow.marks)
print("Original: ",obj.marks)
print("Deep: ",deep.marks)
```
