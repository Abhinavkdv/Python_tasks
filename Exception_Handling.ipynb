```python
# 26
class invalidOperatorError(Exception):
    pass

def check(inp):
    if inp != "+" and inp != "-" and inp != "/" and inp != "*":
        raise invalidOperatorError("Invalid Operator")
    return True

try:
    a = int(input("Enter number:"))
    b = int(input("Enter number:"))
    op = str(input("Enter operator:"))
    # if b==0 and op == '/':
    #     raise ZeroDivisionError("ZeroDivisionError")
    if check(op):
        if op == "+":
            print(a+b)
        elif op == "-":
            print(a-b)
        elif op == "/":
            print(a/b)
        elif op == "*":
            print(a*b)
except (ZeroDivisionError,invalidOperatorError,ValueError) as e:
    print(e)
```

```python
# 28
class BankError(Exception):
    pass
    
class BankSystem:
    def __init__(self,bal):
        self.bal=bal

    def deposit(self,amount):
        if amount < 0:
            raise BankError("deposit amount should be greater than zero")
        else:
            self.bal+=amount

    def withdraw(self,amount):
        if amount > self.bal:
            raise BankError("withdraw amount is greater than balance")

try:
    obj = BankSystem(30_000)
    # obj.deposit(-10_000)
    obj.withdraw(50_000)
except BankError as e:
    print(e)
```
