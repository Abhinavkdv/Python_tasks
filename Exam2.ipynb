# Part 2 — Detect & Fix Data Quality Problems 

```python
# Q6. Duplicate investigation 
print("Number of duplicate rows:",df.duplicated().sum())
```

```python
# Q7. Remove duplicates
df = df.drop_duplicates()
print("Number of duplicate rows After removing duplicates:",df.duplicated().sum())
```

```python
# Q8. Numeric column investigation 
potentially_numeric = ['Age' , 'Annual_Income' , 'Account_Balance' , 'Credit_Score' , 'Monthly_Spend' , 'Transactions']
df.select_dtypes(['int64','float64'])
```

```python
# Q9. Fix Age
df['Age'] = pd.to_numeric(df['Age'],errors = "coerce")
median = df['Age'].median()
print("median of the age in the dataset:",median)
df['Age'] = df['Age'].fillna(median)
```

```python
# Q10. Fix Annual Income 
df['Annual_Income'] = pd.to_numeric(df['Annual_Income'] , errors="coerce")
df['Annual_Income'].astype('float64')
median = df['Annual_Income'].median()
print("median of the Annual Income in the dataset:",median)
df['Annual_Income'] = df['Annual_Income'].fillna(median)
# df['Annual_Income'].head(50)
```

# Part 3 — Numerical Cleaning 

```python
# Q11. Clean Account Balance 
df['Account_Balance'] = pd.to_numeric(df['Account_Balance'] , errors='coerce')
mean = df['Account_Balance'].mean()
print(f"mean of the total account balance is {mean}")
df['Account_Balance'] =df['Account_Balance'].fillna(mean)
# df['Account_Balance'].head(50)
```

```python
# Q12. Clean Credit Score 
df['Credit_Score'] = pd.to_numeric(df['Credit_Score'] , errors='coerce')
median = df['Credit_Score'].median()
print("median of the Credit Score in the dataset:",median)
df['Credit_Score'] = df['Credit_Score'].fillna(median)
df['Credit_Score'] = df['Credit_Score'].astype('int64')
df['Credit_Score'].dtypes
# df['Credit_Score'].head(100)
```

```python
# Q13. Clean Monthly Spend 
df['Monthly_Spend'] = pd.to_numeric(df['Monthly_Spend'],errors="coerce")
median = df['Monthly_Spend'].median()
print("median of the Monthly Spendings in the dataset:",median)
df['Monthly_Spend'] = df['Monthly_Spend'].fillna(median)
df['Monthly_Spend'].dtypes
```

```python
# Q14. Clean Transactions 

df['Transactions'] = pd.to_numeric(df['Transactions'],errors="coerce")
median = df['Transactions'].median()
print("median of the Transactions in the dataset:",median)
df['Transactions'] = df['Transactions'].fillna(median)
df['Transactions'] = df['Transactions'].astype('int64')
df['Transactions'].dtypes
```

```python
# Q15. Clean Satisfaction Score 

df['Satisfaction_Score'] = pd.to_numeric(df['Satisfaction_Score'],errors="coerce")
mean = df['Satisfaction_Score'].median()
print("mean of the Satisfaction Score in the dataset:",mean)
df['Satisfaction_Score'] = df['Satisfaction_Score'].fillna(mean)
df['Satisfaction_Score'] = df['Satisfaction_Score'].astype('float64')
df['Satisfaction_Score'].dtypes
```

# Part 4 — Date Cleaning 

```python
# Q16. Investigate Join_Date
df["Join_Date"].head()
df["Join_Date"].isnull().sum()
```

```python
# Q17. Convert Join_Date 
df["Join_Date"] = pd.to_datetime(df['Join_Date'], yearfirst= True , errors="coerce")
df["Join_Date"].head(50)
```

```python
# Q18. Handle missing dates
median = df['Join_Date'].median()
df['Join_Date'] = df['Join_Date'].fillna(median)
df['Join_Date'].head()
```

# Part 5 — Categorical Cleaning

```python
# Q19. Clean City
df = df.replace({"City" : {"Mumbai":"Mumbai" , "mumbai":"Mumbai",
                          "Hyderabad":"Hyderabad","hyderabad":"Hyderabad",
                          "Delhi":"Delhi","delhi":"Delhi",
                          "Kolkata":"Kolkata" , "kolkata":"Kolkata",
                          "Pune":"Pune" , "pune":"Pune",
                          "Chennai":"Chennai" , "chennai":"Chennai",
                          "Bangalore":"Bangalore" , "bangalore":"Bangalore"}})
df['City'].value_counts()
```

```python
# Q20. Analyze Segment 
print(f"Null values present in the segment column :{df['Segment'].isnull().sum()}")
df['Segment'].value_counts()
df['Segment'] = df['Segment'].fillna('SME')
df['Segment'].value_counts()
```

```python
# Q21. Clean Active 

df = df.replace({"Active" : {"YES":"Yes" , "Yes":"Yes" , "Y":"Yes" , "yes":"Yes" , "1":"Yes" , 
                             "NO":"No" , "No":"No" , "N":"No" , "no":"No" , "0":"No"}})
df['Active'].value_counts()
```

```python
# Q22. Clean Risk_Level 

print(f"Null values present in the Active column :{df['Risk_Level'].isnull().sum()}")
df['Risk_Level'].value_counts()
```
