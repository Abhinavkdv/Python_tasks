```python
# 16
def calculate(a, b, operation):
    if operation == "+":
        print(a+b)
    elif operation == "-":
        print(a-b)
    elif operation == "*":
        print(a*b)
    elif operation == "/":
        print(a/b)
    elif operation == "%":
        print(a%b)
    else:
        print("Invalid operation")
calculate(a=5,b=6,operation="+")
```

```python
# 24
def avg(*args):
    a=0
    s=0
    for i in args:
        a+=i
        s+=1
    return a/s
avg(10,10,10,10,10)
```
